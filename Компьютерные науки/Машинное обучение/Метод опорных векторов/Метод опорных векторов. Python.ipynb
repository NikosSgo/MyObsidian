```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.svm import SVC

X, y = make_blobs(n_samples=30, centers=2, random_state=6)
model = SVC(kernel='linear').fit(X, y)

xx, yy = np.meshgrid(
    np.linspace(X[:, 0].min() - 1, X[:, 0].max() + 1, 100),
    np.linspace(X[:, 1].min() - 1, X[:, 1].max() + 1, 100),
)
Z = model.decision_function(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)

fig, ax = plt.subplots()
ax.scatter(X[:, 0], X[:, 1], c=y, cmap='coolwarm')
ax.contour(xx, yy, Z, levels=[-1, 0, 1], colors='black', linestyles=['--', '-', '--'])
ax.scatter(
    model.support_vectors_[:, 0], model.support_vectors_[:, 1],
    s=180,
    facecolors='none',
    edgecolors='yellow',
    linewidths=3,
)
fig
```

- Сплошная линия — разделяющая гиперплоскость.
- Пунктирные линии — границы максимального зазора.
- Точки с жёлтыми кольцами — опорные векторы `model.support_vectors_`.
