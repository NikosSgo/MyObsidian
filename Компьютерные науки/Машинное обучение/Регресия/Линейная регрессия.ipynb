**Линейная регрессия** - это параметрический метод [[Статистическое обучение|статистического обучения с учителем]] для прогноза числовой целевой переменной. 

Модель предполагает, что среднее значение ответа **линейно** зависит от признаков.

Формула:
$$Y = \beta_0 + \beta_1X_1 + \ldots + \beta_pX_p + \varepsilon.$$
$Y$ - целевая переменная; $X$ - признак; $\beta$ - коэффициенты перед признаками; $\varepsilon$  - [[Разложение ошибки на компоненты#Случайная ошибка. Irreducible Error|случайная ошибка]].

Чаще всего коэффициенты подбирают **МНК**.
То есть минимизирует сумму квадратов остатков, минимизирует, например [[Градиентный спуск|градиентным спуском]] или аналитическим способом.
$$RSS = \sum_{i=1}^{n}(y_i - \hat y_i)^2.$$
Коэффициенты, которые получились при этом хорошо интерпретируются:
- **Абсолютное значение** коэффициента показывает **силу влияния** признака на результат, но требует стандартизации для сравнения.
- Знак коэффициента показывает направление признака. 
	- Минус -> снижение. 
	- Плюс  -> увеличение.

Подходит только для обнаружения **линейных зависимостей**. Имеет смысл, когда корреляция Пирсона между признаками и целевой переменной достаточно велика. При мультиколлинеарности коэффициенты становятся ненадёжными.

### Интерактивный пример в JupyMD

Пусть истинная зависимость имеет вид:

$$
y=3+2x+\varepsilon,
$$

где $\varepsilon$ — случайный шум. Сгенерируем наблюдения, а затем восстановим коэффициенты прямой методом наименьших квадратов.

Ячейки используют общее состояние Python. Их нужно выполнять сверху вниз или запустить все сразу командой `JupyMD: Run all code blocks in current note`.

Создадим выборку:

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)

x = np.linspace(0, 10, 30)
noise = rng.normal(loc=0, scale=2, size=x.size)
y = 3 + 2 * x + noise

print(f"Количество наблюдений: {x.size}")
print("Первые значения x:", np.round(x[:5], 2))
print("Первые значения y:", np.round(y[:5], 2))
```

Для одного признака оценки коэффициентов МНК вычисляются аналитически:

$$
\hat\beta_1=
\frac{\sum_{i=1}^{n}(x_i-\bar x)(y_i-\bar y)}
{\sum_{i=1}^{n}(x_i-\bar x)^2},
$$

$$
\hat\beta_0=\bar y-\hat\beta_1\bar x.
$$

```python
x_mean = x.mean()
y_mean = y.mean()

beta_1 = np.sum((x - x_mean) * (y - y_mean)) / np.sum((x - x_mean) ** 2)
beta_0 = y_mean - beta_1 * x_mean

y_pred = beta_0 + beta_1 * x

print(f"Свободный коэффициент beta_0: {beta_0:.3f}")
print(f"Коэффициент наклона beta_1: {beta_1:.3f}")
print(f"Полученная модель: y = {beta_0:.3f} + {beta_1:.3f}x")
```

Оценим ошибку на обучающей выборке и долю объяснённой дисперсии:

```python
residuals = y - y_pred
rss = np.sum(residuals ** 2)
mse = np.mean(residuals ** 2)
tss = np.sum((y - y.mean()) ** 2)
r_squared = 1 - rss / tss

new_x = 12
new_prediction = beta_0 + beta_1 * new_x

print(f"RSS: {rss:.3f}")
print(f"MSE: {mse:.3f}")
print(f"R²: {r_squared:.3f}")
print(f"Прогноз для x = {new_x}: {new_prediction:.3f}")
```

Построим исходные наблюдения и найденную прямую:

```python
plt.figure(figsize=(8, 5))
plt.scatter(x, y, label="Наблюдения", color="steelblue")
plt.plot(x, y_pred, label="Линейная регрессия", color="crimson", linewidth=2)

plt.xlabel("x")
plt.ylabel("y")
plt.title("Линейная регрессия методом наименьших квадратов")
plt.grid(alpha=0.25)
plt.legend()
plt.show()
```

Из-за случайного шума оценки не обязаны точно совпадать с истинными значениями $\beta_0=3$ и $\beta_1=2$, но при подходящих данных должны быть близки к ним. Повторный запуск первой ячейки воспроизводит ту же выборку, поскольку генератор получает фиксированное начальное значение `42`.
